```python
import pandas as pd
import numpy as np

url = "https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"
df = pd.read_csv(url)
print("Файл загружен")

n = int(input("Введите количество строк n: "))
print("7. Первые", n, "строк:")
print(df.head(n))

print("\n8. Последние 5 строк:")
print(df.tail(5))

print("\n9. Размер таблицы:", df.shape)

print("\n10. Столбцы:")
print(df.columns.tolist())

print("\n11. Типы данных:")
print(df.dtypes)

chisla = df.select_dtypes(include=np.number).columns.tolist()
slova = df.select_dtypes(include='object').columns.tolist()
print("\n12. Числовые:", chisla)
print("Категориальные:", slova)

print("\n13. Статистика:")
print(df.describe())

print("\n14. Пассажиры по классам:")
print(df['Pclass'].value_counts())

print("\n15. Уникальные порты Embarked:")
print(df['Embarked'].unique())
print("Количество:", df['Embarked'].nunique())

print("\n16. Пропуски:")
print(df.isnull().sum())

print("\n17. Процент пропусков:")
print((df.isnull().sum() / len(df)) * 100)

print("\n18. Больше всего пропусков в:")
print(df.isnull().sum().idxmax())

print("\n19. Возраст ДО:")
age_before = df['Age'].describe()
print(age_before)

mediana = df['Age'].median()
df['Age'] = df['Age'].fillna(mediana)
print("\n20. Пропуски в Age заполнены медианой:", mediana)

print("\n21. Возраст ПОСЛЕ:")
age_after = df['Age'].describe()
print(age_after)

print("\n22. Сравнение:")
tablica = pd.DataFrame({
    'Показатель': ['Среднее', 'Медиана', 'Минимум', 'Максимум', 'Ст.отклонение', 'Кол-во известных'],
    'До': [age_before['mean'], age_before['50%'], age_before['min'], age_before['max'], age_before['std'], age_before['count']],
    'После': [age_after['mean'], age_after['50%'], age_after['min'], age_after['max'], age_after['std'], age_after['count']]
})
print(tablica)

df = df.dropna(subset=['Embarked'])
print("\n23. Удалены строки без Embarked, осталось:", len(df))

cabin_procent = (df['Cabin'].isnull().sum() / len(df)) * 100
print("\n24. Доля пропусков в Cabin:", round(cabin_procent, 2), "%")
if cabin_procent > 50:
    df = df.drop(columns=['Cabin'])
    print("Удалил Cabin")

print("\n25. Остатки пропусков:")
print(df.isnull().sum())

print("\n26. Новый размер таблицы:", df.shape)

surv = df["Survived"].to_numpy()
print("\n31. Процент выживших:", round(np.mean(surv) * 100, 2), "%")

muzh = df[df['Sex'] == 'male']
zhen = df[df['Sex'] == 'female']
muzh_surv = np.mean(muzh['Survived'].to_numpy()) * 100
zhen_surv = np.mean(zhen['Survived'].to_numpy()) * 100
print("32. Выжили мужчины:", round(muzh_surv, 2), "%, женщины:", round(zhen_surv, 2), "%")

print("33. Средний возраст мужчин:", round(np.mean(muzh['Age'].to_numpy()), 2), ", женщин:", round(np.mean(zhen['Age'].to_numpy()), 2))

m_v = np.mean(muzh[muzh['Survived'] == 1]['Age'].to_numpy())
m_p = np.mean(muzh[muzh['Survived'] == 0]['Age'].to_numpy())
w_v = np.mean(zhen[zhen['Survived'] == 1]['Age'].to_numpy())
w_p = np.mean(zhen[zhen['Survived'] == 0]['Age'].to_numpy())

print("34. Мужчины: выжили", round(m_v, 2), ", погибли", round(m_p, 2))
print("    Женщины: выжили", round(w_v, 2), ", погибли", round(w_p, 2))

itog = pd.DataFrame({
    'Группа': ['Мужчины выжившие', 'Мужчины погибшие', 'Женщины выжившие', 'Женщины погибшие'],
    'Средний возраст': [round(m_v, 2), round(m_p, 2), round(w_v, 2), round(w_p, 2)]
})
print("\nТаблица сравнительного анализа:")
print(itog)

fare = df['Fare'].to_numpy()
print("\n35. Билет: мин =", round(np.min(fare), 2), ", макс =", round(np.max(fare), 2), ", средняя =", round(np.mean(fare), 2))

age = df['Age'].to_numpy()
print("36. Ст. отклонение возраста:", round(np.std(age, ddof=1), 2))

print("\nЗадание 1. Мужчины > 30 лет в 1 классе:")
task1 = df[(df['Sex'] == 'male') & (df['Age'] > 30) & (df['Pclass'] == 1)]
print("Найдено:", len(task1))
print(task1.head(10))

task2 = df[(df['Age'] < 18) | ((df['Sex'] == 'female') & (df['Survived'] == 1))]
print("\nЗадание 2. Дети или выжившие женщины:", len(task2))

sred_bilet = df[df['Pclass'] == 3]['Fare'].mean()
task3 = df[(df['Pclass'] == 3) & (df['Fare'] > sred_bilet)]
print("Задание 3. 3 класс с дорогим билетом:", len(task3))

task4 = df[(df['Age'] >= 20) & (df['Age'] <= 40) & (df['Embarked'] == 'S')]
print("Задание 4. 20-40 лет из S:", len(task4))

group1 = df.groupby(['Pclass', 'Sex']).agg(
    Количество=('PassengerId', 'count'),
    Средний_возраст=('Age', 'mean'),
    Доля_выживших=('Survived', 'mean'),
    Средняя_стоимость=('Fare', 'mean')
)
group1['Доля_выживших'] = group1['Доля_выживших'] * 100
print("\n37. Статистика по классам и полу:")
print(group1)

group2 = df.groupby('Embarked').agg(
    Количество=('PassengerId', 'count'),
    Процент_выживших=('Survived', 'mean')
)
group2['Процент_выживших'] = group2['Процент_выживших'] * 100
print("\n38. Статистика по портам:")
print(group2)

print("\n39. Класс с наибольшей долей выживших:")
print(df.groupby('Pclass')['Survived'].mean().idxmax())

df.to_csv("titanic_processed.csv", index=False)
group1.to_csv("titanic_statistics.csv")

df_check = pd.read_csv("titanic_processed.csv")
print("\n42. Проверка. Размер сохраненного файла:", df_check.shape)
print("Файлы сохранены")
```

# Ответы на контрольные вопросы и выводы

## Контрольные вопросы

**Почему Pclass - категориальный признак?**
Числа 1, 2, 3 обозначают класс билета (люкс, средний, эконом), а не количество. Арифметика с ними не имеет смысла, т. к. это просто ярлыки категорий.

**Вопросы 27-30:**

**Вопрос 27.** Медиана предпочтительнее среднего, так как она устойчива к выбросам, например, к пассажирам преклонного возраста.

**Вопрос 28.** Изменились: среднее, стандартное отклонение и количество известных значений. Не изменились: минимум, максимум и медиана.

**Вопрос 29.** Стандартное отклонение изменилось, потому что добавили 177 одинаковых значений (медиану), и разброс данных уменьшился.

**Вопрос 30.** Исходное распределение не сохраняется, появляется искусственный «пик» на значении медианы.

**Различия между мужчинами и женщинами?**
Женщины выживали кардинально чаще мужчин (около 74% против 18%).

**Отличие «И» и «ИЛИ» при фильтрации?**
«И» (&) требует выполнения всех условий сразу. «ИЛИ» (|) требует выполнения хотя бы одного условия.

**Отличие группировки от фильтрации?**
Фильтрация отбирает строки по условию. Группировка объединяет строки в группы и позволяет считать сводные метрики (среднее, количество) для каждой группы.
